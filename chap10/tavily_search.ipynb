```python
from langchain_community.tools import TavilySearchResults

tavily_search = TavilySearchResults(
    max_results=5
)
```

```python
res = tavily_search.invoke({"query":"2026 한국 경제 전망"})
for r in res:
    print(r)
```

```python
from IPython.display import JSON
JSON(res).data
```

```python
from tavily import TavilyClient

client = TavilyClient()

content = client.search(
    "2026년 한국 경제 전망",
    search_depth="advanced",
    include_raw_content=True,
)

JSON(content).data
```

```python
from langchain_classic.adapters.openai import convert_openai_messages
from langchain_openai import ChatOpenAI


llm = ChatOpenAI(model="gpt-4o-mini")

# 질문에 대한 자료를 TavilyClient로 가져오기
query = "앞으로의 컴퓨터 주요 부품 값 전망"

content = client.search(
    query,
    include_raw_content=True,
    search_depth="advanced"
)["results"]
```

```python
# 프롬프트 설정
prompt = [{
    "role" : "system",
    "content" : (
        "당신은 신문 기자를 쓰는 기자 AI입니다. \n"
        "당신은 주어진 정보를 바탕으로 객관적이고 체계적으로 작성된 기사를 써야 합니다. \n"
    )
}, {
    "role" : "user",
    "content": (
        f'정보: """{content}"""\n\n'
        f'위의 정보를 사용하여, 다음 질문에 대해 자세한 보고서를 한국어로 작성하세요.: "{query}\n'
        '-신문 기사 형식을 사용하되, MLA 표준을 준수하는 markdown 문법을 사용해주세요.'
        '-활용한 자료는 출처를 명시하세요.'
    )
    
}]
```

```python
report = llm.invoke(prompt).content
print(report)
```
